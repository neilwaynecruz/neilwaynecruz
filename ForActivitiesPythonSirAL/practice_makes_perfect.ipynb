```python
import csv
from os import path
from tabulate import tabulate

file_path = "C:\\Users\\neilw\\pythonJupyter\\ForActivitiesPythonSirAL\\csvfiles\\GradingSystem.csv"


def computeGrades(q1, q2, q3, q4, grading_system):
    gs = grading_system.upper()

    if gs == "A":
        return (int(q1) + int(q2) + int(q3) + int(q4)) / 4
    elif gs == "C":
         return int(q4)
    elif gs == "W":
        return ((int(q1) * 0.25 )+ (int(q2) * 0.20) + (int(q3) * 0.25) + (int(q4) * 0.40))
    else:
        return 0


def addRec():
    mode = 'a' if path.exists(file_path) else 'w'
    
    with open(file_path,mode,newline="") as fp:
        fprint = csv.writer(fp)

        stud_num = input("Student Number: ")
        last_name = input("Last Name: ")
        first_name = input("First Name: ")
        q1 = input("Quiz 1: ")
        q2 = input("Quiz 2: ")
        q3 = input("Quiz 3: ")
        q4 = input("Quiz 4: ")
        grading_system = input("Grading system [A - Average/ C - Based only on last quiz/ W - Weighted Average]: ").upper()

        fprint.writerow([stud_num,last_name,first_name,q1,q2,q3,q4,grading_system])

    print("Record Added!\n")

def viewRec():
    if not path.exists(file_path):
        print("No records found.\n")
        return

    with open(file_path,"r") as fp:
        fscan = csv.reader(fp)
        print(f"{'Student #':<12}{'Last Name':<15}{'First Name':<20}{'Q1':<6}{'Q2':<6}{'Q3':<6}{'Q4':<6}{'GS':<6}{'Grade':<6}")

        rows = []
        for row in fscan:
            stud_num, last_name, first_name, q1,q2,q3,q4,grading_system = row
            computed_grade = computeGrades(q1,q2,q3,q4,grading_system)
            print(f"{stud_num:<12}{last_name:<15}{first_name:<20}{q1:<6}{q2:<6}{q3:<6}{q4:<6}{grading_system:<6}{computed_grade:<6.2f}")
            # rows.append([stud_num, last_name, first_name, q1, q2, q3, q4, grading_system, f"{computed_grade:.2f}"])
        
        
        print()
        # print(tabulate(rows, headers=["Student #","Last Name","First Name","Q1","Q2","Q3","Q4","GS","Grade"], tablefmt="rounded_outline"))

def searchStud():
    if not path.exists(file_path):
        print("No records found.\n")
        return

    search_stud = input("Student number to search: ")
    found = False

    with open(file_path,"r",newline="") as fp:
        fscan = csv.reader(fp)
        

        for row in fscan:
            stud_num, last_name, first_name, q1,q2,q3,q4,grading_system = row

            if stud_num == search_stud:
                print(f"{'Student #':<12}{'Last Name':<15}{'First Name':<20}{'Q1':<6}{'Q2':<6}{'Q3':<6}{'Q4':<6}{'GS':<6}{'Grade':<6}")
                computed_grade = computeGrades(q1,q2,q3,q4,grading_system)
                print(f"{stud_num:<12}{last_name:<15}{first_name:<20}{q1:<6}{q2:<6}{q3:<6}{q4:<6}{grading_system:<6}{computed_grade:<6.2f}")
                found = True
                break
        
        print()

    if not found:
        print("Student not found.\n")


try:
    while True:
        print("Press 1: Add new student records.")
        print("Press 2: View all student records with computed grades.")
        print("Press 3: Search a student by Student #.")
        print("Press 4: Exit.")

        choice = input("Enter your choice: ")
        print()

        if choice.isdigit():
            if int(choice) == 1:
                addRec()
            elif int(choice) == 2:
                viewRec()
            elif int(choice) == 3:
                searchStud()
            elif int(choice) == 4:
                break
            else:
                print("Invalid choice. Please try again.\n")
        else:
            print("Please enter a number (1–4).\n")

except Exception as e:
    print(f"Error Message: {e}")
```
